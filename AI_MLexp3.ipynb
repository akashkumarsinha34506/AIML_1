# TIC TAC TOE GAME EXP

```python
import random

board = [" " for i in range(9)]


def display_board():
    print()
    print(board[0], "|", board[1], "|", board[2])
    print("--+---+--")
    print(board[3], "|", board[4], "|", board[5])
    print("--+---+--")
    print(board[6], "|", board[7], "|", board[8])
    print()


def check_winner(symbol):
    wins = [
        [0,1,2], [3,4,5], [6,7,8],
        [0,3,6], [1,4,7], [2,5,8],
        [0,4,8], [2,4,6]
    ]

    for w in wins:
        if board[w[0]] == symbol and board[w[1]] == symbol and board[w[2]] == symbol:
            return True

    return False


def computer_move():

    # First check if computer can win
    for i in range(9):
        if board[i] == " ":
            board[i] = "O"

            if check_winner("O"):
                return

            board[i] = " "

    # Block the user if they are about to win
    for i in range(9):
        if board[i] == " ":
            board[i] = "X"

            if check_winner("X"):
                board[i] = "O"
                return

            board[i] = " "

    # Choose a random empty position
    empty_positions = []

    for i in range(9):
        if board[i] == " ":
            empty_positions.append(i)

    if len(empty_positions) > 0:
        computer_position = random.choice(empty_positions)
        board[computer_position] = "O"


print("TIC TAC TOE")
print("You = X")
print("Computer = O")

display_board()


for turn in range(9):

    # USER MOVE
    while True:
        position = int(input("Enter position (1-9): ")) - 1

        if position >= 0 and position <= 8 and board[position] == " ":
            board[position] = "X"
            break
        else:
            print("Invalid position! Try again.")

    display_board()

    # Check user win
    if check_winner("X"):
        print("You Win!")
        break

    # Check draw after user's move
    if " " not in board:
        print("It's a Draw!")
        break

    # COMPUTER MOVE
    computer_move()

    print("Computer moved:")
    display_board()

    # Check computer win
    if check_winner("O"):
        print("Computer Wins!")
        break

    # Check draw after computer's move
    if " " not in board:
        print("It's a Draw!")
        break
```
