```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

print("All libraries imported successfully!")
```

```python
data = {
    "Name": ["Amit", "Rahul", "Priya", "Sneha", "Akash",
             "Neha", "Rohan", "Pooja", "Karan", "Anjali"],
    "Marks": [85, 78, 92, 88, 95, 72, 80, 90, 65, 86],
    "Study_Hours": [4, 3, 6, 5, 7, 2, 4, 6, 1, 5]
}

df = pd.DataFrame(data)

print("Student Dataset:")
print(df)

print("\nAverage Marks:", df["Marks"].mean())
print("Highest Marks:", df["Marks"].max())
print("Lowest Marks:", df["Marks"].min())
```

1. Box plot

```python
plt.figure(figsize=(7, 5))

sns.boxplot(y=df["Marks"])

plt.title("Box Plot of Student Marks")
plt.ylabel("Marks")
plt.show()
```

2. Pie chart

```python
plt.figure(figsize=(7, 7))

plt.pie(
    df["Marks"],
    labels=df["Name"],
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Percentage Distribution of Student Marks")
plt.show()
```

3. Line graph

```python
plt.figure(figsize=(8, 5))

plt.plot(
    df["Name"],
    df["Marks"],
    marker="o",
    linestyle="-"
)

plt.title("Student Marks - Line Graph")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

4. Scatter plot

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    x=df["Study_Hours"],
    y=df["Marks"],
    s=100
)

plt.title("Study Hours vs Student Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.grid(True)
plt.show()
```

5. Histogram

```python
plt.figure(figsize=(8, 5))

plt.hist(
    df["Marks"],
    bins=5,
    edgecolor="black"
)

plt.title("Distribution of Student Marks")
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.grid(axis="y")
plt.show()
```
