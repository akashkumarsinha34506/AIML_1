# Breadth first search

```python
from collections import deque
#creating a graph

graph={
    'A':['B','C'],
    'B':['D','E'],
    'C':['F','G'],
    'D':['H'],
    'E':['I'],
    'F':[],
    'G':[],
    'H':[],
    'I':[]
}
#CREATING A FUNCTION

def bfs(graph,start):
  visited=[]
  queue=deque()

  visited.append(start)
  queue.append(start)

  while queue:#it means that unless the queue is empty this while loop will run
     node =queue.popleft()
     print(node, end="")

     #visit all neighbour
     for neighbour in graph[node]:
       if neighbour not in visited:
         visited.append(neighbour)
         queue.append(neighbour)


start=input("enter the starting node:").upper()
print("Following is the Breadth First Search")

bfs(graph,start)
```

```python
from collections import deque

#creating a graph

graph={
    '1':['2','3','4'],
    '2':['5','6'],
    '3':[],
    '4':[],
    '5':['7'],
    '6':['8'],
    '7':[],
    '8':[]

}

#creating a function
def bfs(graph,start):
  visited=[]
  queue=deque()

  visited.append(start)
  queue.append(start)

  while queue:
    node=queue.popleft()
    print(node,end="")

    for neighbour in graph[node]:
      if neighbour not in visited:
        visited.append(neighbour)
        queue.append(neighbour)



print("this the demostration of DFS")
start=input("enetr the stating node:")
bfs(graph,start)

```
